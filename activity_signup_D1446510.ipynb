```markdown
# HW09 兩場活動報名整理器
本作業主要運用 Python 的字串方法、容器型別（tuple, list, set, dict）與 f-string 格式化來整理活動報名名單、進行集合運算並計算應收費用。
```

````python
# ==========================================
# 測試一：原始資料
# ==========================================
print("=== 測試一 ===")

# 1. 原始資料定義
activity_names = ("桌遊體驗", "電影欣賞")  # tuple：保存固定且不可變的活動名稱
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}  # dict：以活動名稱為鍵查詢每人單價

# 2. 讀取活動名稱與整理串列
# 從 activity_names 拆包讀取活動名稱
act_board, act_movie = activity_names

# 將原始文字移除半形空白、轉成小寫，再依逗號拆成 list
board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

# 3. 使用集合進行去重與比較名單
board_set = set(board_list)  # 桌遊去重集合
movie_set = set(movie_list)  # 電影去重集合

# 集合運算：
# 交集 (&)：兩場都有參加的共同參加者
common_attendees = sorted(list(board_set & movie_set))
# 聯集 (|)：至少參加一場的全部不重複參加者
all_attendees = sorted(list(board_set | movie_set))
# 差集 (-)：只參加桌遊者（桌遊有、電影沒有）
only_board = sorted(list(board_set - movie_set))
# 差集 (-)：只參加電影者（電影有、桌遊沒有）
only_movie = sorted(list(movie_set - board_set))

# 4. 建立排序串列並用切片取得前兩位
preview_list = all_attendees[:2]

# 5. 建立字典保存摘要與計算費用
summary = {
    "桌遊人數": len(board_set),
    "電影人數": len(movie_set),
    "共同人數": len(common_attendees),
    "總不重複人數": len(all_attendees),
    "報名人次": len(board_set) + len(movie_set), # 兩場去重人數相加（參加兩場者算兩人次）
    "桌遊應收": len(board_set) * unit_fees[act_board],
    "電影應收": len(movie_set) * unit_fees[act_movie]
}
summary["總應收"] = summary["桌遊應收"] + summary["電影應收"]

# 6. 格式化輸出完整摘要
print("HW09 活動報名摘要")
print(f"活動：{act_board}，每人 {unit_fees[act_board]:.2f} 元")
print(f"活動：{act_movie}，每人 {unit_fees[act_movie]:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{sorted(list(board_set))}")
print(f"電影去重名單：{sorted(list(movie_set))}")
print(f"桌遊人數：{summary['桌遊人數']} 人")
print(f"電影人數：{summary['電影人數']} 人")
print(f"共同參加者：{common_attendees}")
print(f"共同人數：{summary['共同人數']} 人")
print(f"只參加桌遊者：{only_board}")
print(f"只參加電影者：{only_movie}")
print(f"全部參加者：{all_attendees}")
print(f"總不重複人數：{summary['總不重複人數']} 人")
print(f"報名人次：{summary['報名人次']} 人次")
print(f"通知名單前兩位：{preview_list}")
print(f"桌遊應收：{summary['桌遊應收']:.2f} 元")
print(f"電影應收：{summary['電影應收']:.2f} 元")
print(f"總應收：{summary['總應收']:.2f} 元")
````
